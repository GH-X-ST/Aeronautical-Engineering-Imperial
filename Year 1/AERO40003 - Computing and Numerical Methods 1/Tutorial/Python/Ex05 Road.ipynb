Two cars make a journey from city A to city B by different routes. The speed of Car 1 at certain times is given in the table below:

|Time (min) |0.0|1.0 |10.0| 11.0|50.0 |60.0|80.0|85.0|
|-----------|---|----|----|-----|-----|----|----|----|
|Car1 (km/h)|0.0|50.0|50.0|110.0|100.0|46.0|35.0|0.0 |

The speed of Car 2 is given as follows:

|Time (min) |0.0|2.0 |15.0| 18.0|80.0 |85.0|100.0|101.0|
|-----------|---|----|----|-----|-----|----|-----|-----|
|Car2 (km/h)|0.0|30.0|40.0|90.0 |90.0 |20.0|35.0 |0.0  |


It can be assumed that velocity changes linearly between time points (i.e. piecewise-constant acceleration).

## Task 1
Create a class `Car` which encapsulates the velocity data above for one vehicle. Store the velocity data as member data. Write a function `SetData` which adds time/velocity measurements using a suitable container. Create a `CalculateDistanceTravelled()` function, which calculates the distance travelled using the data points in the class.

```python
class Car:
    def __init__(self,t,v):
        self.t=t
        self.v=v
        
    def SetData(self,T,V):
        t=self.t
        v=self.v
        t.extend(T)
        v.extend(V)
        
    def CalculateDistanceTravelled(self):
        D=0
        t=self.t
        v=self.v
        T=[]
        V=[]
        self.SetData(T,V)
        for i in range (0,len(t)-1):
            D+=0.5*(v[i+1]+v[i])*(t[i+1]-t[i-1])
        return D
```
