### Task 1
Write code to compute the first $N=20$ terms of the Fibonacci sequence, given by the recurrence relation:
$$x_n = x_{n-1} + x_{n-2} \quad 0<n<N,$$
with $x_0 = x_1 = 1$.

```python
# Write a function for the Fibonacci sequence
def FibonacciSequence(N):
    a=1
    b=1
    x=a+b
    print(a,end=',')
    print(b,end=',')
    while x<=N:
        a=b
        b=x
        print(x,end=',')
        x=a+b
```

```python
# compute the first 𝑁=20 terms of the Fibonacci sequence
FibonacciSequence(20)
```

### Task 2
Write Python code which computes a list of the prime numbers up to $N=1000$.

```python
# Write a function computes a list of the prime numbers 
def Prime(N):
    i=2
    for i in range(2,N):
        j=2
        for j in range(2,i):
            if (i%j)==0:
                break
        else:
            print(i,end=',')
```

```python
# Compute a list of the prime number up to N=1000
Prime(1000)
```

### Task 3
Write Python code to generate a list of 101 random numbers and calculate their minimum, maximum, mean, median and standard deviation.

**Hint**: You can write functions to compute these yourself, or have a look at the packages:
* *statistics*: https://docs.python.org/3/library/statistics.html
* *random*: https://docs.python.org/3/library/random.html

```python
# Write Python function to generate a list of random numbers
def rand(N):
    import random
    r=[]
    for i in range(1,N+1):
        a=random.random()
        r.append(a)
    print(r)
    # Calculte the min and max of numbers
    min=r[0]
    max=r[0]
    for j in range(0,len(r)):
        if min>=r[j]:
            min=r[j]
        else:
            pass
    for k in range(0,len(r)):
        if max<=r[k]:
             max=r[k]
        else:
            pass
    print('Minimum of the random numbers is' ,min)
    print('Maximum of the random numbers is' ,max)  
    # Calculte the mean and median of numbers
    import statistics
    me=statistics.mean(r)
    med=statistics.median(r)
    print('Mean of the random numbers is' ,me)
    print('Median of the random numbers is' ,med)
    # Calculte the standard deviation of numbers
    st=0
    for l in range(0,len(r)):
        st+=(r[l]-me)**2
        st=(st/len(r))**0.5
    print('Standard deviation of the random numbers is' ,st)
```

```python
rand(101)
```

### Task 4
Write a function to find the set of all duplicated numbers in a list of integers.

Generate a list of 15 random integers between 0 and 9 inclusive to test your function. Print out the original list and the found set of random integers to check your result.

```python
# Write a function to find the set of all duplicated numbers in a list of integers
def FindDuplicated(A):
    D=[]
    for i in range(0,len(A)-1):
        for j in range(i+1,len(A)-1):
            if A[i]==A[j]:
                D.append(A[i])
                D.append(A[j])
                break
        else:
            pass
    print(D)
```

```python
# Generate random integers between 0 and 9
import random
R=[]
for i in range(1,16):
    R.append(random.randint(0,9))
print(R)
```

```python
# Find the duplicated numbers
FindDuplicated(R)
```
